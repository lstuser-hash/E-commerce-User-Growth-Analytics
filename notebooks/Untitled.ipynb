```python
orders["order_purchase_timestamp"] = pd.to_datetime(
    orders["order_purchase_timestamp"]
)

orders.info()
```

```python
orders["order_month"] = (
    orders["order_purchase_timestamp"]
    .dt.to_period("M")
)
```

```python
monthly_orders = (
    orders
    .groupby("order_month")
    .size()
)

monthly_orders.head()
```

```python
monthly_growth = monthly_orders.pct_change()

monthly_growth.head()
```

```python
payments = pd.read_csv(
    "../data/raw/olist_order_payments_dataset.csv"
)

payments.head()
```

```python
payments.shape
```

```python
order_payment = orders.merge(
    payments,
    on="order_id",
    how="inner"
)

order_payment.head()
```

```python
user_orders = order_payment.merge(
    customers[
        [
            "customer_id",
            "customer_unique_id"
        ]
    ],
    on="customer_id",
    how="inner"
)

user_orders.head()
```

```python
analysis_date = (
    user_orders["order_purchase_timestamp"]
    .max()
    +
    pd.Timedelta(days=1)
)

analysis_date
```

```python
rfm = user_orders.groupby(
    "customer_unique_id"
).agg(
    {
        "order_purchase_timestamp":
        lambda x:
        (analysis_date-x.max()).days,

        "order_id":
        "count",

        "payment_value":
        "sum"
    }
)
```

```python
rfm.columns = [
    "Recency",
    "Frequency",
    "Monetary"
]

rfm.head()
```

```python
user_orders.head()
```

```python
user_orders["order_purchase_timestamp"] = pd.to_datetime(
    user_orders["order_purchase_timestamp"]
)
```

```python
analysis_date = (
    user_orders["order_purchase_timestamp"].max()
    +
    pd.Timedelta(days=1)
)

analysis_date
```

```python
rfm = user_orders.groupby(
    "customer_unique_id"
).agg(
    {
        "order_purchase_timestamp":
        lambda x:
        (analysis_date - x.max()).days,

        "order_id":
        "count",

        "payment_value":
        "sum"
    }
)
```

```python
rfm.columns = [
    "Recency",
    "Frequency",
    "Monetary"
]

rfm.head()
```

```python
rfm.describe()
```

## RFM Analysis Insights

Key findings:

1. The dataset contains 96K+ unique customers.
2. Most customers only made one purchase, indicating a significant opportunity for improving customer retention.
3. Customer spending shows high variation, suggesting the need for customer segmentation.
4. A small group of high-value customers may contribute disproportionately to revenue.

```python
rfm["R_score"] = pd.qcut(
    rfm["Recency"],
    5,
    labels=[5,4,3,2,1]
)
```

```python
rfm["F_score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    5,
    labels=[1,2,3,4,5]
)
```

```python
rfm["M_score"] = pd.qcut(
    rfm["Monetary"],
    5,
    labels=[1,2,3,4,5]
)
```

```python
rfm.head()
```

```python
def segment_customer(score):
    if score >= 13:
        return "Champions"
    elif score >= 10:
        return "Loyal Customers"
    elif score >= 7:
        return "Potential Customers"
    elif score >= 4:
        return "At Risk"
    else:
        return "Lost Customers"


rfm["Customer_Segment"] = rfm["RFM_Score"].apply(
    segment_customer
)

rfm.head()
```

```python
segment_count = (
    rfm["Customer_Segment"]
    .value_counts()
)

segment_count
```

```python
plt.figure(figsize=(10,6))

sns.barplot(
    x=segment_count.values,
    y=segment_count.index
)

plt.title(
    "Customer Segmentation Based on RFM"
)

plt.xlabel(
    "Number of Customers"
)

plt.ylabel(
    "Customer Segment"
)

plt.show()
```

```python
rfm.to_csv(
    "../data/processed/rfm_customer_segments.csv",
    index=True
)
```
